```python
img_bgr = cv2.imread('churras.jpg')
```

```python
img_grey = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY) #converting bgr to grey
img_grey
```

```python
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB) #converting bgr to rgb
img_rgb


```

```python
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV) #converting bgr to hsv
img_hsv
```

```python
plt.imshow(img_grey, cmap='gray')
plt.show()
```

```python
plt.figure(figsize=(5,3))#500x300
plt.imshow(img_rgb)
plt.title('BBQ IN RGB')
plt.show()
```

```python
plt.subplot(2,2,1)
plt.imshow(img_grey, cmap='gray')
plt.title('BBQ GREY')

plt.subplot(2,2,2)
plt.imshow(img_rgb)
plt.title('BBQ RGB')

plt.subplot(2,2,3)
plt.imshow(img_hsv)
plt.title('BBQ HSV')

plt.subplot(2,2,4)
plt.imshow(img_bgr)
plt.title('BBQ BGR')

plt.show()

plt
```

```python
fig, ax = plt.subplots(2,2,figsize=(6,4))

#eixo
ax[0][0].imshow(img_rgb) #img1
ax[0][0].set_title('BBQ  RGB')
ax[0][0].axis('off')

ax[0][1].imshow(img_bgr) #img2
ax[0][1].set_title('BBQ  BGR')
ax[0][1].axis('off')

ax[1][0].imshow(img_hsv) #img3
ax[1][0].set_title('BBQ  HSV')
ax[1][0].axis('off')

ax[1][1].imshow(img_grey, cmap='gray') #img4
ax[1][1].set_title('BBQ  GREY')
ax[1][1].axis('off')

plt.tight_layout()
plt.savefig('matriz_BBQ.png')
plt.show()
```
